```python
movies = data['results']
```

```python
movie_list = []
for movie in movies:
    movie_list.append({
        "title": movie["title"],
        "rating": movie["vote_average"],
        "popularity": movie["popularity"],
        "overview": movie["overview"],
        "release_date": movie["release_date"]
    })
```

```python
import pandas as pd

df = pd.DataFrame(movie_list)
df.head()
```

```python
df['overview'] = df['overview'].fillna('')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['overview'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend(movie_title):
    index = df[df['title'] == movie_title].index[0]
    
    distances = similarity[index]
    
    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movie_list:
        print(df.iloc[i[0]].title)
```

```python
recommend("Malena")
```

```python
import requests
import pandas as pd

API_KEY = "9af4f73fa1d57fda800a2ccddd43fed7"

url = f"https://api.themoviedb.org/3/movie/popular?api_key={API_KEY}"

response = requests.get(url)

data = response.json()

movies = data['results']

movie_list = []

for movie in movies:

    genre_ids = movie.get('genre_ids', [])

    movie_list.append({
        "movie_id": movie["id"],
        "title": movie["title"],
        "overview": movie["overview"],
        "rating": movie["vote_average"],
        "popularity": movie["popularity"],
        "release_date": movie["release_date"],
        "poster_path": movie["poster_path"],
        "genre_ids": genre_ids
    })

df = pd.DataFrame(movie_list)

df.head()
```

```python
df.to_csv("../data/movies_v2.csv", index=False)
```

```python
genre_dict = {
    28: "Action",
    12: "Adventure",
    16: "Animation",
    35: "Comedy",
    80: "Crime",
    99: "Documentary",
    18: "Drama",
    10751: "Family",
    14: "Fantasy",
    36: "History",
    27: "Horror",
    10402: "Music",
    9648: "Mystery",
    10749: "Romance",
    878: "Science Fiction",
    10770: "TV Movie",
    53: "Thriller",
    10752: "War",
    37: "Western"
}
```

```python
def convert_genres(genre_ids):

    genre_names = []

    for genre_id in genre_ids:

        if genre_id in genre_dict:
            genre_names.append(genre_dict[genre_id])

    return " ".join(genre_names)

df['genres'] = df['genre_ids'].apply(convert_genres)

df[['title', 'genres']].head()
```

```python
df['combined_features'] = (
    df['title'].fillna('') + ' ' +
    df['overview'].fillna('') + ' ' +
    df['genres'].fillna('')
)

df[['title', 'combined_features']].head()
```

```python
df.to_csv("../data/movies_v2.csv", index=False)
```
